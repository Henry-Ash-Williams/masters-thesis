```python
import requests
import numpy as np
from rich import inspect
from bs4 import BeautifulSoup
import bs4
from bs4 import UnicodeDammit
import re
from tqdm import tqdm
import urllib
import pandas as pd

base_url = "https://www.portablefreeware.com"
url = f"{base_url}/index.php?s=100&p={{}}"
n_pages = 20
```

```python
def get(url: str):
    res = requests.get(url)
    return BeautifulSoup(UnicodeDammit(res.content).unicode_markup)
```

```python
def try_download(url):
    try:
        res = requests.get(url)
        content = BeautifulSoup(res.content)
    except:
        print(f"Failed to get or parse {url}")
        return
    executable_links = content.find_all("a", href=re.compile(".+\/.+\.exe"))
    urls = []
    for link in executable_links:
        try:
            href = link.attrs["href"]
        except:
            continue
        if not href.startswith("http"):
            href = f"{res.url}/{href}"
        urls.append(href)
    return urls
```

```python
def try_get_download_links(url: str, page_id: int):
    url = url.format(page_id)
    print(f"Requesting {url}")
    content = get(url)
    entries = content.find_all(class_="entry group nomargin")
    print(f"Found {len(entries)} potential downloads")

    to_download = []
    for entry in entries:
        downloads = list(entry.find("li", class_="download").children)
        for download in downloads:
            if isinstance(download, bs4.element.NavigableString):
                continue

            if "32" in download.string or "|" in download.string:
                continue

            to_download.append(f"{base_url}/{download.attrs['href']}")

    download_links = []
    for this_url in to_download:
        download_links.append(try_download(this_url))

    return [
        str(np.random.choice(link, 1)[0])
        for link in download_links
        if link is not None and len(link) > 0
    ]
```

```python
from concurrent.futures import ThreadPoolExecutor


def fetch_links(page: int) -> list[str]:
    return try_get_download_links(url, page)


with ThreadPoolExecutor() as executor:
    results = executor.map(fetch_links, range(1, n_pages + 1))

links = [link for result in results for link in result]

pd.DataFrame(links).to_csv("benign-links.csv")
```
