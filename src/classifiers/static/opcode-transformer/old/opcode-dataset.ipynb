```python
import json
from tqdm import tqdm
from collections import Counter

DISASM_PATH = "/Volumes/malware-dataset/obfuscated-benign/disassembly/"
```

```python
from json import JSONDecodeError
import os


def process_single_instruction(instr):
    return instr.split(" ")[0]


def process_file(path):
    with open(path, "r") as f:
        content = json.load(f)

        if content == []:
            raise Exception("Empty Sequence")

        return [process_single_instruction(instr) for instr in content]


items = [
    os.path.join(DISASM_PATH, file)
    for file in os.listdir(DISASM_PATH)
    if file.endswith(".json")
]
item_iterator = tqdm(items)

for path in item_iterator:
    basename = path.split("/")[-1]

    if basename.startswith("._"):
        continue

    item_iterator.set_postfix(file=basename)

    try:
        instrs = process_file(path)
        if isinstance(instrs, JSONDecodeError):
            raise Exception("Could not decode this file")
    except Exception as e:
        item_iterator.write(f"ERROR: {e}")
        continue

    out_path = os.path.join(
        DISASM_PATH, f'opcode-sequences/{basename.replace(".json", ".txt")}'
    )

    with open(out_path, "w") as file:
        file.write(" ".join(instrs))
```
