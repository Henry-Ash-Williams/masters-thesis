```python
import json
from tqdm import tqdm
from collections import Counter

LABELS_PATH = "/Volumes/New Volume/malware-detection-dataset/opcodes/labels.json"

labels = json.load(open(LABELS_PATH, "r"))
instruction_frequency = Counter()
```

```python
def process_single_instruction(instr):
    return instr.split(" ")[0]


def process_file(path):
    content = json.load(open(path, "r"))

    if content == []:
        raise Exception("Empty Sequence")

    return [process_single_instruction(instr) for instr in content]


new_labels = {}

for path, label in tqdm(labels.items()):
    if path.split("/")[-1].startswith("._"):
        continue
    try:
        instrs = process_file(path)
    except:
        continue

    instruction_frequency.update(instrs)

    out_path = (
        path.replace("/data-2/", "/processed-data/")
        .replace(".json", ".txt")
        .replace(".exe", "")
    )

    new_labels[out_path] = label

    with open(out_path, "w") as file:
        file.write(" ".join(instrs))

with open(
    "/Volumes/New Volume/malware-detection-dataset/opcodes/processed-data/labels.json",
    "w",
) as file:
    json.dump(new_labels, file)
```
