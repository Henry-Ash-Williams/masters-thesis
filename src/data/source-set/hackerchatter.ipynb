```python
import pandas as pd 
import numpy as np

from collections import Counter

df = pd.read_csv("/Users/henrywilliams/Documents/uni/masters-thesis/src/data/Hackerchatter.org.csv")
windows_df = df[['windows' in p.lower() if not isinstance(p, float) else False for p in df['Platform'] ]]

windows_df
```

# Family Analysis

```python
family = Counter() 

for row in windows_df.itertuples():
    families = row.Family

    if isinstance(families, float):
        continue

    families = families.split(",")
    family.update([f.strip() for f in families])

family
```

# Language Analysis

```python
languages = Counter([row.Language.lower() for row in windows_df.itertuples() if not isinstance(row.Language, float)]) 
languages
```

# Topic Analysis

```python
topic = Counter() 

for row in windows_df.itertuples():
    topics = row.Topic

    if isinstance(topics, float):
        continue

    topics = topics.split(",")
    topic.update([t.strip() for t in topics])

topic
```
