```python
import requests
import bs4
import random

BASE_URL = "https://portableapps.com"
res = requests.get(f"{BASE_URL}/apps")
content = bs4.BeautifulSoup(res.content)
```

```python
lis = [
    li
    for directory in content.find_all(class_="appdirectory")
    for li in directory.find_all("li")
]

links = [BASE_URL + li.find("a").attrs["href"] for li in lis]
len(links)
```

```python
link = random.choice(links)
import urllib.parse


download_link = (
    bs4.BeautifulSoup(requests.get(link).content)
    .find(class_="download-box")
    .find("a")
    .attrs["href"]
    .split("/")
)
query = urllib.parse.parse_qs(download_link[-1])


# safe_string = BASE_URL + urllib.parse.quote_plus()
# requests.get(safe_string)
# print(download_link)
```
