```python
from os import PathLike
import os
from typing import List
import subprocess
import utils

dataset = utils.data.MalwareDataset()


def extract_strings(file_path: PathLike, number=4) -> List[str]:
    if not os.path.exists(file_path):
        raise FileNotFoundError("File does not exist")

    strings = subprocess.run(["strings", f"-{number}", file_path], capture_output=True)

    return strings.stdout.decode("ascii").splitlines()


strings = extract_strings(dataset[0][0])
```

```python
from collections import Counter
import numpy as np
from nltk.corpus import brown


def compute_bigram_frequencies(text: str) -> dict:
    text = text.lower()
    bigram_counts = Counter(zip(text, text[1:]))
    total_bigrams = sum(bigram_counts.values())

    return {pair: count / total_bigrams for pair, count in bigram_counts.items()}


def log_likelihood(s: str) -> float:
    s = [c for c in s.lower() if c.isalpha()]
    s_bigrams = zip(s, s[1:])
    log_l = sum(np.log(BIGRAMS[bigram]) for bigram in s_bigrams)
    return log_l


def is_likely_natural(s, threshold: float = -50) -> bool:
    return log_likelihood(s) > threshold


BIGRAMS = {(chr(a), chr(b)): 1e-6 for a in range(97, 123) for b in range(97, 123)}

BIGRAMS.update(compute_bigram_frequencies(brown.raw()))
```

```python
import string
import pandas as pd


def ratio(s: str):
    if len(s) == 0:
        return (0.0, 0.0, 0.0)
    num_alpha = len([c for c in s if c.isalpha()])
    num_numeric = len([c for c in s if c.isnumeric()])
    num_sym = len([c for c in s if c in string.punctuation])
    total = len(s)

    return (num_alpha / total, num_numeric / total, num_sym / total)


strings = np.array(strings)
arr = np.array([list(ratio(s)) for s in strings])
strings[arr.max(axis=1) > np.sum(arr[:, 1:], axis=1)]
```
